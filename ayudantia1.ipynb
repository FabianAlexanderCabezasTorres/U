```python
#include <stdio.h>

int main(){
    int c, ele;
    printf("meteme el pico rico oh si yea")
    printf("Ingresa una opción (1-5) y luego presiona ENTER: ");
    ele = getchar() - '0';
    getchar(); // Esto limpia el salto de línea (Enter) de la memoria

    printf("Ingresa un número: ");
    if((c = getchar()) != EOF) {
        printf("El valor antes de operar es: %d\n", c);
    }

    if(ele == 5){
        c = c ^ 0b01010101
    } else if(ele == 4){
        c = c - "a"
    }

    printf("El valor final es: ");
    if (ele == 4 || ele == 5){
        printf("%c\n", c)
    } else {
        printf("%d\n", c)
    }

    return 0;
}
```
