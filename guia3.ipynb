Ejercicio 1: Escribe una función que tome un entero positivo e indique
cuántos bits 1 contiene su representación binaria.


```python
#include <stdio.h>

int contar(int numero){
    int contador = 0;

    while (numero > 0) {
        contador += numero & 1;
        numero = numero >> 1;
    }
    return contador
}

int main(){
    int numeroprueba = 13;
    int resultado;

    resultado = contar(numeroprueba);

    printf("el numero %d en binario tiene %d bits encendidos.",numeroprueba,resultado);
    return 0;
}
```

Guia 5 Ejercicio 1: Cálculo de la Altura de un Árbol Binario
Dado un árbol binario que almacena valores enteros, escriba un programa en C que
determine la altura máxima de la estructura. La altura se define como la cantidad total
de niveles desde el nodo raíz hasta la hoja más distante. En caso de que la estructura
se encuentre completamente vacía, la función correspondiente debe retornar el valor
cero.


  • Caso Base: Si el puntero recibido es nulo, la rama no aporta niveles y retorna
cero.

  • Caso Recursivo: Calcula recursivamente la profundidad de ambas ramas.
Compara cuál de las dos es más profunda y le suma uno para contabilizar el
nivel actual.


```python
int calcular_altura(struct Nodo *nodo) {
    // 1. Caso Base: El freno de emergencia
    if (nodo == NULL) {
        return 0;
    }

    // 2. Caso Recursivo: Bajar por las ramas
    int izq = calcular_altura(nodo->izquierdo);
    int der = calcular_altura(nodo->derecho);

    // 3. Comparar y sumar el nivel actual
    if (izq > der) {
        return izq + 1;
    } else {
        return der + 1;
    }
}
```
